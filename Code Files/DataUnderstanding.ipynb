# IMDB Movies

## 1. Introduction

- This project focuses on scraping, cleaning, and analyzing IMDb title data to understand trends related to genres, ratings, certificates, runtime, metascores, popularity, and release years.
- The dataset includes both movies and television content collected from IMDb genre-based search pages.

The project involves:

1. Web scraping using Selenium and BeautifulSoup
2. Data cleaning and preprocessing
3. Exploratory Data Analysis (EDA)
4. Univariate and Bivariate Analysis
5. Visualizations and statistical interpretation

## 2. Search for Relevant Websites 
- Find the websites which are allowing you to scrape the data from the domain/use case you have chosen.


To build the dataset, a suitable website that provides structured information about movies was required. After evaluating a few entertainment and movie-related websites, IMDb was selected as the primary data source.

 ### IMDb

- IMDb is one of the most widely used online movie databases and contains comprehensive information such as  titles, release years, genres, ratings, vote counts, runtime, content certificates, and metascores.
-  The website presents this information in a consistent format, which makes it suitable for web scraping and subsequent data analysis.
- Data was collected from ten genres, including Action, Comedy, Drama, Romance, Thriller, Crime, Adventure, Sci-Fi, Animation, and Fantasy.

The following attributes were extracted from each listing:

1. Title
2. Year
3. Genre
4. Rating
5. Votes
6. Runtime
7. Certificate
8. Metascore

IMDb was chosen because:

- It provides a large and diverse collection of movie records.
- Most of the required attributes are displayed directly on the search results pages.
- The website structure is relatively consistent, making automated extraction feasible.
- The data is suitable for performing exploratory data analysis and deriving meaningful insights.

This website served as the primary source for constructing the dataset used throughout the project.

### Problem Statement

- With the rapid growth of digital entertainment platforms, a vast number of movies and television shows are released every year.
- These titles differ in genre, duration, popularity, critic reviews, and audience reception.
-  Understanding how these characteristics relate to audience ratings can provide valuable insights into viewer preferences and content performance.
- This project aims to perform Exploratory Data Analysis (EDA) on IMDb title data to identify patterns, trends, and relationships among different features and to determine which factors are associated with higher IMDb ratings and greater popularity.

### Objective of the Study

The primary objectives of this project are:

- To analyze entertainment titles collected from IMDb, including both movies and television content such as TV series, mini-series, and animated shows.
- To analyze the distribution of titles across genres and release years.
- To examine how IMDb ratings vary with popularity (votes), runtime, and metascore.
- To compare audience ratings across different genres.
- To study the relationship between critic scores and audience ratings.
- To uncover trends in both movies and television content.

### Target Feature

The main target feature selected for this project is: **Rating (IMDb Rating)**

IMDb Rating is used as the target feature because it represents audience evaluation and serves as a key indicator of the success and reception of a title.

### Input Features

The following features are considered relevant to the target feature:

- Year – Release year of the title.
- Genre – One or more categories associated with the title.
- Votes – Number of user ratings, representing popularity.
- Runtime – Duration in minutes.
- Certificate – Content classification such as TV-MA, TV-14, PG-13, and R.
- Metascore – Critic score provided by Metacritic.

### Scope of the Dataset

- The dataset includes both movies and television titles.
- This was identified from the presence of certificate categories such as TV-MA, TV-PG, and TV-14, which are commonly used for television programming.
-  Therefore, the term title is used throughout the project to represent all forms of entertainment content rather than only theatrical films.

### Business Relevance

- Insights from this analysis can help content creators, streaming platforms, and entertainment analysts better understand the characteristics associated with highly rated and popular titles.
- Such findings may support decisions related to content development, audience targeting, and market analysis.
